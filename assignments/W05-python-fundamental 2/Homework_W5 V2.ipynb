```python
# Bài 20
import math
a = float(input("Nhập hệ số a: "))
b = float(input("Nhập hệ số b: "))
c = float(input("Nhập hệ số c: "))
if a == 0:
    if b == 0:
        if c == 0:
            print("Phương trình có vô số nghiệm.")
        else:
            print("Phương trình vô nghiệm.")
    else:
      x = -c / b
      print(f"Phương trình bậc nhất có nghiệm x = {x:.2f}")
else:
  delta = b**2 - 4 * a * c
  if delta < 0:
        print("Phương trình vô nghiệm (trong tập số thực).")
  elif delta == 0:
        x = -b / (2 * a)
        print(f"Phương trình có nghiệm kép x = {x:.2f}")
  else:
    x1 = (-b + math.sqrt(delta)) / (2 * a)
    x2 = (-b - math.sqrt(delta)) / (2 * a)
    print(f"Phương trình có 2 nghiệm phân biệt:")
    print(f"x1 = {x1:.2f}")
    print(f"x2 = {x2:.2f}")

```

```python
# Bài 33
import math
x0, y0, R = map(float, input("Nhập x0, y0, R: ").split())

xa, ya = map(float, input("Nhập xa, ya: ").split())

d = math.sqrt((xa - x0)**2 + (ya - y0)**2)

EPS = 1e-9

if abs(d - R) < EPS:
    print("Điểm A nằm trên đường tròn")
elif d < R:
    print("Điểm A nằm bên trong đường tròn")
else:
    print("Điểm A nằm bên ngoài đường tròn")
```
