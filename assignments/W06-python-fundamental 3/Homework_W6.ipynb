```python
# Bài 19
def giai_bat_phuong_trinh_bac_1(a, b):
    # Trường hợp a > 0: Giữ nguyên chiều dấu
    if a > 0:
        x = -b / a
        return f"Bất phương trình có nghiệm: x >= {x}"
    # Trường hợp a < 0: Đổi chiều dấu bất phương trình
    elif a < 0:
        x = -b / a
        return f"Bất phương trình có nghiệm: x <= {x}"
    # Trường hợp a = 0: Dựa vào giá trị của b
    else:
        # Nếu b >= 0 (ví dụ: 3 >= 0 luôn đúng với mọi x)
        if b >= 0:
            return "Bất phương trình có vô số nghiệm!"
        # Nếu b < 0 (ví dụ: -3 >= 0 là vô lý)
        else:
            return "Bất phương trình vô nghiệm!"
a = float(input("Nhập hệ số a: "))
b = float(input("Nhập hệ số b: ")) 
print(f"Kết quả {giai_bat_phuong_trinh_bac_1(a, b)}")
```

```python
# Bài 20
import math
def giai_phuong_trinh_bac_2(a, b, c):
    # Trường hợp đặc biệt nếu a = 0 thì trở thành phương trình bậc nhất: bx + c = 0
    if a == 0:
        if b == 0:
            if c == 0:
                return "Phương trình có vô số nghiệm!"
            else:
                return "Phương trình vô nghiệm!"
        else:
            x = -c / b
            return f"Phương trình có một nghiệm (bậc 1): x = {x}"
    
    # Tính Delta (Δ) = b^2 - 4ac
    delta = (b ** 2) - (4 * a * c)
    # Δ < 0: Vô nghiệm
    if delta < 0:
        return "Phương trình vô nghiệm!"
    # Δ = 0: Nghiệm kép
    elif delta == 0:
        x = -b / (2 * a)
        return f"Phương trình có nghiệm kép: x = {x}"
    # Δ > 0: 2 nghiệm phân biệt
    else:
        x1 = (-b + math.sqrt(delta)) / (2 * a)
        x2 = (-b - math.sqrt(delta)) / (2 * a)
        return f"Phương trình có 2 nghiệm phân biệt:\nx1 = {x1}\nx2 = {x2}"
a = float(input("Nhập hệ số a: "))
b = float(input("Nhập hệ số b: "))
c = float(input("Nhập hệ số c: "))
print(f"Kết quả {giai_phuong_trinh_bac_2(a, b, c)}")
```

```python
# Bài 32
def giai_he_phuong_trinh (a, b, m, c, d, n):
    # Tính định thức D
    D = a * d - b * c
    if D == 0:
        return "Hệ phương trình vô nghiệm hoặc vô số nghiệm"
    else:
        # Tính nghiệm x và y
        x = (m * d - b * n) / D
        y = (a * n - m * c) / D
        return f"hệ phương trình có 1 cặp nghiệm: x = {x}, y = {y}"
a = float(input("Nhập hệ số a: "))
b = float(input("Nhập hệ số b: "))
m = float(input("Nhập hệ số m: "))
c = float(input("Nhập hệ số c: "))
d = float(input("Nhập hệ số d: "))
n = float(input("Nhập hệ số n: "))
print(giai_he_phuong_trinh(a, b, m, c, d, n))
```

```python
# Bài 38
def tinh_giai_thua(n):
    if n < 0:
        return "Không thể tính giai thừa của số âm"
    elif n == 0 or n == 1:
        return 1
    else:
        giai_thua = 1
        for i in range(2, n + 1):
            giai_thua *= i
        return giai_thua

n = int(input("Nhập số n: "))
print(f"{n} giai thừa = {tinh_giai_thua(n)}")
```

```python
# Bài 40
def tinh_tong_chia_het_3_hoac_5(m, n):
    tong = 0
    for i in range(m, n + 1):
        if i % 3 == 0 or i % 5 == 0:
            tong += i
    return tong

m = int(input("Nhập số m: "))
n = int(input("Nhập số n: "))
print(f"kết quả = {tinh_tong_chia_het_3_hoac_5(m, n)}")
```

```python
# Bài 43
def dem_uoc_cua_n(n):
    if n < 0:
        return "Không thể đếm ước số của số âm"
    elif n == 0:
        return "Số 0 có vô số ước số"
    else:
        dem = 0
        for i in range(1, n + 1):
            if n % i == 0:
                dem += 1
        return dem
n = int(input("Nhập số n: "))
print(f"{n} có {dem_uoc_cua_n(n)} ước")
```

```python
# Bài 44
def dem_so_trong_khoang (danh_sach):
    dem = 0
    for so in danh_sach:
        if 10 <= so <= 20:
            dem += 1
    return dem
n = int(input("Nhập số lượng phần tử trong danh sách: "))
danh_sach = []
for i in range(n):
    so = float(input(f"Nhập phần tử thứ {i + 1}: "))
    danh_sach.append(so)
print(f"các số >10 và <20 là: {dem_so_trong_khoang(danh_sach)}")
```

```python
# Bài 45
import math
def kiem_tra_so_nguyen_to(n):
    if n < 2:
        return False
    if n == 2:
        return True
    if n % 2 == 0:
        return False
    for i in range(3, int(math.sqrt(n)) + 1, 2):
        if n % i == 0:
            return False
    return True
n = int(input("Nhập số n: "))
if kiem_tra_so_nguyen_to(n):
    print(f"{n} là số nguyên tố")
else:
    print(f"{n} không phải là số nguyên tố")
```

```python
# Bài 46
def kiem_tra_so_hoan_hao(n):
    if n <= 0:
        return False
    tong_uoc = 0
    for i in range(1, n):
        if n % i == 0:
            tong_uoc += i
    return tong_uoc == n
n = int(input("Nhập số n: "))
if kiem_tra_so_hoan_hao(n):
    print(f"{n} là số hoàn hảo")
else:
    print(f"{n} không phải là số hoàn hảo")
```

```python
# Bài 47
def kiem_tra_cac_so_hoan_hao (n):
    so_hoan_hao = []
    for i in range(1, n + 1):
        if kiem_tra_so_hoan_hao(i):
            so_hoan_hao.append(i)
    return so_hoan_hao
n = int(input("Nhập số n: "))
print (f"{kiem_tra_cac_so_hoan_hao(n)}")
```

```python
# Bài 48
def kiem_tra_so_chinh_phuong(n):
    if n < 0:
        return False
    can_bac_hai = int(n ** 0.5)
    return can_bac_hai * can_bac_hai == n

n = int(input("Nhập số n: "))
if kiem_tra_so_chinh_phuong(n):
    print(f"{n} là số chính phương")
else:
    print(f"{n} không phải là số chính phương")
```
