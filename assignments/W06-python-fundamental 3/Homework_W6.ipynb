```python
# Bài 1
def tinh_so_ban(i):
    # Nhập sĩ số của lớp từ bàn phím
    return i-1

i = int(input("Lớp bạn có bao nhiêu người? "))
so_ban = tinh_so_ban(i)
print(f"Vậy bạn có {so_ban} người bạn trong lớp")
```

```python
# Bài 2
def tinh_toan(a, b):
    tong = a + b
    hieu = a - b
    tich = a * b
    thuong = a / b if b != 0 else "Không thể chia cho 0"
    return tong, hieu, tich, thuong

a = int(input("Nhập a: "))
b = int(input("Nhập b: "))
tong, hieu, tich, thuong = tinh_toan(a, b)
print(f"Tong 2 so: {tong:.2f}")
print(f"Hieu 2 so: {hieu:.2f}")
print(f"Tich 2 so: {tich:.2f}")
print(f"Thương 2 so: {thuong}")
```

```python
# Bài 5
def tinh_diem_thi_dua(a, b, c, t, n, m):
    tong_diem_bi_tru = (a * t) + (b * n) + (c * m)
    return tong_diem_bi_tru
a, b, c, t, n, m = map(int, input().split())
print (f"Tổng điểm bị trừ: {tinh_diem_thi_dua(a, b, c, t, n, m)} ")
```

```python
def tinh_tien_dien(chiso1, chiso2):
    if chiso2 < chiso1:
        return "Lỗi: Chỉ số tháng này không được nhỏ hơn chỉ số tháng trước!"
    so_kw = chiso2 - chiso1
    tien = 0
    # Tính tiền điện theo bậc thang
    if so_kw <= 60:
        tien = so_kw * 5
    elif so_kw <= 160:
        tien = 60 * 5 + (so_kw - 60) * 8
    else:
        tien = 60 * 5 + 100 * 8 + (so_kw - 160) * 10
    return tien
ho_ten = input("Họ và tên: ")
chiso1, chiso2 = map(int, input("Nhập chỉ số tháng trước, chỉ số tháng này: ").split())
print(f"Họ và tên: {ho_ten}")
print(f"Số tiền điện phải trả: {tinh_tien_dien(chiso1, chiso2):.2f}")
```

```python
# Bài 10
def tach_chu_so(n):
    hang_tram = n // 100
    hang_chuc = (n // 10) % 10
    hang_don_vi = n % 10
    return hang_tram, hang_chuc, hang_don_vi
n = int(input("Nhập số n: "))
hang_tram, hang_chuc, hang_don_vi = tach_chu_so(n)
print(f"Chữ số hàng trăm: {hang_tram}")
print(f"Chữ số hàng chục: {hang_chuc}")
print(f"Chữ số hàng đơn vị: {hang_don_vi}")
```

```python
# Bài 12
def tinh_s_p_hinh_chu_nhat (a, b):
    s = a * b
    p = 2 * (a + b)
    return s, p

a = float(input("Nhập độ dài cạnh hình chữ nhật: "))
b = float(input("Nhập độ rộng cạnh hình chữ nhật: "))
s, p = tinh_s_p_hinh_chu_nhat(a, b)
print(f"Diện tích hình chữ nhật: {s}")
print(f"Chu vi hình chữ nhật: {p}")   
```

```python
# Bài 13
import math
def tinh_s_p_hinh_tron (r):
    s = math.pi * r * r
    p = 2 * math.pi * r
    return s, p
r = float(input("Nhập bán kính hình tròn: "))
s, p = tinh_s_p_hinh_tron(r)
print(f"Diện tích hình tròn: {s:.2f}")
print(f"Chu vi hình tròn: {p:.2f}")
```

```python
# Bài 14
import math
def tinh_s_p_hinh_tam_giac (a, b, c):
  if (a > 0 and b > 0 and c > 0) and (a + b > c) and (a + c > b) and (b + c > a):
        chu_vi = a + b + c
        p = chu_vi / 2
        s = math.sqrt(p * (p - a) * (p - b) * (p - c))
        return True, chu_vi, s
  else:
        return False, 0, 0
a = float(input("Nhập độ dài cạnh thứ nhất: "))
b = float(input("Nhập độ dài cạnh thứ hai: "))
c = float(input("Nhập độ dài cạnh thứ ba: "))
hop_le, chu_vi, s = tinh_s_p_hinh_tam_giac(a, b, c)
if hop_le:
    print(f"Diện tích hình tam giác: {s:.2f}")
    print(f"Chu vi hình tam giác: {chu_vi:.2f}")
else:
    print(f"{a}, {b}, {c} không phải 3 cạnh của một tam giác.")
```

```python
# Bài 18
def giai_phuong_trinh_bac_1(a, b):
    # Trường hợp a khác 0: Phương trình có một nghiệm duy nhất
    if a != 0:
        x = -b / a
        return f"Phuong trình có một nghiệm! x = {x}"
    # Trường hợp a bằng 0
    else:
        # Nếu cả a và b đều bằng 0: Vô số nghiệm
        if b == 0:
            return "Phuong trình có vô số nghiệm!"
        # Nếu a = 0 nhưng b khác 0: Vô nghiệm
        else:
            return "Phuong trình vô nghiệm!"
a = float(input("Nhập hệ số a: "))
b = float(input("Nhập hệ số b: "))
print(f"Kết quả {giai_phuong_trinh_bac_1(a, b)}")
```

```python
# Bài 19
def giai_bat_phuong_trinh_bac_1(a, b):
    # Trường hợp a > 0: Giữ nguyên chiều dấu
    if a > 0:
        x = -b / a
        return f"Bất phương trình có nghiệm: x >= {x}"
    # Trường hợp a < 0: Đổi chiều dấu bất phương trình
    elif a < 0:
        x = -b / a
        return f"Bất phương trình có nghiệm: x <= {x}"
    # Trường hợp a = 0: Dựa vào giá trị của b
    else:
        # Nếu b >= 0 (ví dụ: 3 >= 0 luôn đúng với mọi x)
        if b >= 0:
            return "Bất phương trình có vô số nghiệm!"
        # Nếu b < 0 (ví dụ: -3 >= 0 là vô lý)
        else:
            return "Bất phương trình vô nghiệm!"
a = float(input("Nhập hệ số a: "))
b = float(input("Nhập hệ số b: ")) 
print(f"Kết quả {giai_bat_phuong_trinh_bac_1(a, b)}")
```

```python
# Bài 20
import math
def giai_phuong_trinh_bac_2(a, b, c):
    # Trường hợp đặc biệt nếu a = 0 thì trở thành phương trình bậc nhất: bx + c = 0
    if a == 0:
        if b == 0:
            if c == 0:
                return "Phương trình có vô số nghiệm!"
            else:
                return "Phương trình vô nghiệm!"
        else:
            x = -c / b
            return f"Phương trình có một nghiệm (bậc 1): x = {x}"
    
    # Tính Delta (Δ) = b^2 - 4ac
    delta = (b ** 2) - (4 * a * c)
    # Δ < 0: Vô nghiệm
    if delta < 0:
        return "Phương trình vô nghiệm!"
    # Δ = 0: Nghiệm kép
    elif delta == 0:
        x = -b / (2 * a)
        return f"Phương trình có nghiệm kép: x = {x}"
    # Δ > 0: 2 nghiệm phân biệt
    else:
        x1 = (-b + math.sqrt(delta)) / (2 * a)
        x2 = (-b - math.sqrt(delta)) / (2 * a)
        return f"Phương trình có 2 nghiệm phân biệt:\nx1 = {x1}\nx2 = {x2}"
a = float(input("Nhập hệ số a: "))
b = float(input("Nhập hệ số b: "))
c = float(input("Nhập hệ số c: "))
print(f"Kết quả {giai_phuong_trinh_bac_2(a, b, c)}")
```
