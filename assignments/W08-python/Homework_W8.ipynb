```python
#Bài 1
def tim_cac_so():
    ket_qua = []
    for i in range(2000, 3201):
        if i % 7 == 0 and i % 5 != 0:
            ket_qua.append(str(i))
    return ",".join(ket_qua)
print(tim_cac_so())
```

```python
# Bài 2
def tinh_giai_thua (n):
  if n == 0:
    return 1
  else:
    return n * tinh_giai_thua(n - 1)
n = int(input("nhập số cần tính giai thừa:"))
print (tinh_giai_thua (n))
```

```python
# Bài 9
import datetime
def tinh_tuoi(nam, thang, ngay):
    hom_nay = datetime.date.today()
    tuoi = hom_nay.year - nam
    if (hom_nay.month, hom_nay.day) < (thang, ngay):
        tuoi = tuoi - 1
    return tuoi
s = input("Nhập ngày tháng năm sinh: ")
nam, thang, ngay = s.split("/")
print("Ngày sinh vừa nhập:", s)
print("Số tuổi là:", tinh_tuoi(int(nam), int(thang), int(ngay)))
    
```

```python
# Bài 10
class tinhluong:
    def __init__(self, so_gio, tien_cong_gio):
        self.so_gio = so_gio
        self.tien_cong_gio = tien_cong_gio
        self.gio_chuan = 44
    def tinh_toan(self):
        if self.so_gio < self.gio_chuan:
            return self.so_gio * self.gio_chuan
        else:
            gio_ot = self.so_gio - self.gio_chuan
            return (self.gio_chuan * self.tien_cong_gio) + (gio_ot * self.tien_cong_gio * 1.5)
gio = float(input("nhập số giờ làm việc trong tháng:"))
tien = float(input("nhập giá 1 giờ làm việc:"))
nv = tinhluong(gio,tien)
print(f"tiền lương tháng này là:{nv.tinh_toan()}")
    
```

```python
# Bài 11
import math
class tinhQ:
    def __init__ (self, d, c=50, h=30):
        self.c = c
        self.h = h
        self.d = [float(x.strip()) for x in d.split(",")]
    def tinh_toan (self):
        q_values = []
        for d in self.d:
            q =math.sqrt((2*self.c * d)/self.h)
            q_values.append(str(round(q)))
        return ",".join(q_values)
chuoi_nhap = input ("nhập dãy giá trị d:")
ket_qua= tinhQ(chuoi_nhap).tinh_toan()
print(ket_qua)
```

```python
# Bài 12
def tao_2d(x,y):
    return [[i * j for j in range(y)] for i in range(x)]
nhap_x_y = input("nhập giá trị x,y:")
x,y = map(int, nhap_x_y.split(','))
print (tao_2d(x,y))
```

```python
# Bài 13
def sap_xep_chuoi_tu(chuoi_tu):
    danh_sach_tu = [tu.strip() for tu in chuoi_tu.split(',')]
    danh_sach_tu.sort()
    return ','.join(danh_sach_tu)
nhap_lieu = input("Nhập danh sách các từ, phân tách bằng dấu phẩy: ")
ket_qua = sap_xep_chuoi_tu(nhap_lieu)
print(sap_xep_chuoi_tu(nhap_lieu))
```

```python
# Bài 14
class chuyenchuinhoa:
    def __init__(self):
        self.cac_dong = []
    def nhap_du_lieu(self):
        print("Nhập các dòng văn bản (Để trống rồi nhấn Enter để kết thúc):")
        while True:
            dong = input()
            if dong == "":
                break
            self.cac_dong.append(dong)
    def in_ket_qua(self):
        print("\n--- Kết quả (Chữ in hoa) ---")
        for dong in self.cac_dong:
            print(dong.upper())
app = chuyenchuinhoa()
app.nhap_du_lieu()
app.in_ket_qua()
```

```python
class So:
    def __init__(self, gia_tri):
        self.gia_tri = gia_tri

    def __add__(self, other):
        return So(self.gia_tri + other.gia_tri)
    
    def __str__(self):
        return str(self.gia_tri)

a = So(10)
b = So(20)
c = So(30)
d = a + b + c

print(d)  # 30
```
